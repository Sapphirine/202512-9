## 4. Define Advanced RL Environment

```python
import gymnasium as gym
from gymnasium import spaces

class ESGPortfolioEnv(gym.Env):
    """
    Advanced Gym environment for ESG-driven portfolio optimization.

    State Space:
        - Current portfolio weights (n_assets)
        - Recent returns (lookback * n_assets)
        - Recent volatility (n_assets)
        - ESG scores (n_assets)
        - Portfolio ESG score (1)
        - Portfolio value (1)

    Action Space:
        - Continuous: new portfolio weights [0, 1] for each asset

    Reward:
        - Portfolio return
        + ESG bonus (higher ESG = higher bonus)
        - Risk penalty (higher volatility = higher penalty)
        - Transaction cost penalty
        - ESG constraint penalty (if below threshold)
    """

    metadata = {'render_modes': ['human']}

    def __init__(self, returns_data, esg_scores,
                 initial_balance=1000000,
                 esg_weight=0.3,
                 risk_penalty=0.5,
                 transaction_cost=0.001,
                 min_esg_threshold=50,
                 lookback_window=20):

        super(ESGPortfolioEnv, self).__init__()

        # Data
        self.returns_data = returns_data.values
        self.dates = returns_data.index
        self.esg_scores = esg_scores
        self.n_assets = len(esg_scores)
        self.n_steps = len(returns_data)

        # Hyperparameters
        self.initial_balance = initial_balance
        self.esg_weight = esg_weight
        self.risk_penalty = risk_penalty
        self.transaction_cost = transaction_cost
        self.min_esg_threshold = min_esg_threshold
        self.lookback_window = lookback_window

        # Action space: portfolio weights [0, 1] for each asset
        self.action_space = spaces.Box(
            low=0, high=1, shape=(self.n_assets,), dtype=np.float32
        )

        # State space dimension calculation
        state_dim = (
            self.n_assets +  # current weights
            self.n_assets * self.lookback_window +  # recent returns
            self.n_assets +  # recent volatility
            self.n_assets +  # ESG scores
            1 +  # portfolio ESG score
            1    # normalized portfolio value
        )

        self.observation_space = spaces.Box(
            low=-np.inf, high=np.inf, shape=(state_dim,), dtype=np.float32
        )

        # Episode tracking
        self.current_step = 0
        self.portfolio_value = initial_balance
        self.portfolio_weights = np.array([1.0/self.n_assets] * self.n_assets)
        self.portfolio_history = []
        self.esg_history = []
        self.trades_history = []

    def reset(self, seed=None, options=None):
        """Reset environment to initial state."""
        super().reset(seed=seed)

        self.current_step = self.lookback_window
        self.portfolio_value = self.initial_balance
        self.portfolio_weights = np.array([1.0/self.n_assets] * self.n_assets)
        self.portfolio_history = [self.initial_balance]
        self.esg_history = []
        self.trades_history = []

        return self._get_observation(), {}

    def _get_observation(self):
        """Construct observation/state vector."""
        obs = []

        # Current portfolio weights
        obs.extend(self.portfolio_weights)

        # Recent returns (flattened lookback window)
        start_idx = max(0, self.current_step - self.lookback_window)
        recent_returns = self.returns_data[start_idx:self.current_step]
        if len(recent_returns) < self.lookback_window:
            padding = np.zeros((self.lookback_window - len(recent_returns), self.n_assets))
            recent_returns = np.vstack([padding, recent_returns])
        obs.extend(recent_returns.flatten())

        # Recent volatility
        volatility = np.std(recent_returns, axis=0)
        obs.extend(volatility)

        # ESG scores (normalized)
        obs.extend(self.esg_scores / 100.0)

        # Portfolio ESG score
        portfolio_esg = np.dot(self.portfolio_weights, self.esg_scores)
        obs.append(portfolio_esg / 100.0)

        # Normalized portfolio value
        obs.append(self.portfolio_value / self.initial_balance)

        return np.array(obs, dtype=np.float32)

    def step(self, action):
        """Execute one step in the environment."""
        # Normalize action to valid portfolio weights (sum to 1)
        action = np.clip(action, 0, 1)
        new_weights = action / (np.sum(action) + 1e-8)

        # Calculate transaction costs
        turnover = np.sum(np.abs(new_weights - self.portfolio_weights))
        transaction_cost = turnover * self.transaction_cost * self.portfolio_value

        # Get today's returns
        if self.current_step < self.n_steps:
            today_returns = self.returns_data[self.current_step]
        else:
            today_returns = np.zeros(self.n_assets)

        # Calculate portfolio return
        portfolio_return = np.dot(new_weights, today_returns)

        # Update portfolio value
        old_value = self.portfolio_value
        self.portfolio_value = self.portfolio_value * (1 + portfolio_return) - transaction_cost

        # Calculate reward components

        # 1. Return reward (daily return)
        return_reward = portfolio_return * 100  # Scale up for better learning

        # 2. ESG reward
        portfolio_esg = np.dot(new_weights, self.esg_scores)
        esg_reward = (portfolio_esg / 100.0) * self.esg_weight

        # 3. ESG constraint penalty (severe penalty if below threshold)
        if portfolio_esg < self.min_esg_threshold:
            esg_penalty = (self.min_esg_threshold - portfolio_esg) / 100.0 * 2.0
        else:
            esg_penalty = 0

        # 4. Risk penalty (penalize high volatility)
        recent_returns = self.returns_data[max(0, self.current_step-20):self.current_step+1]
        portfolio_volatility = np.std([np.dot(new_weights, r) for r in recent_returns])
        risk_penalty_value = portfolio_volatility * self.risk_penalty * 10

        # 5. Transaction cost penalty
        tc_penalty = (transaction_cost / old_value) * 10  # Normalize

        # Total reward
        reward = return_reward + esg_reward - esg_penalty - risk_penalty_value - tc_penalty

        # Update state
        self.portfolio_weights = new_weights
        self.current_step += 1

        # Track history
        self.portfolio_history.append(self.portfolio_value)
        self.esg_history.append(portfolio_esg)
        self.trades_history.append(turnover)

        # Check if episode is done
        done = self.current_step >= self.n_steps - 1
        terminated = done
        truncated = False

        # Additional info for logging
        info = {
            'portfolio_value': self.portfolio_value,
            'portfolio_return': portfolio_return,
            'portfolio_esg': portfolio_esg,
            'turnover': turnover,
            'transaction_cost': transaction_cost
        }

        obs = self._get_observation()

        return obs, reward, terminated, truncated, info

    def render(self):
        """Render the environment (optional)."""
        print(f"Step: {self.current_step}")
        print(f"Portfolio Value: ${self.portfolio_value:,.2f}")
        print(f"Portfolio ESG: {np.dot(self.portfolio_weights, self.esg_scores):.2f}")

print("ESG Portfolio Environment defined")
```

## 5. Create Training and Validation Environments

```python
# Split data into train and validation sets (80-20 split)
train_size = int(0.8 * len(returns_matrix))

train_returns = returns_matrix.iloc[:train_size]
val_returns = returns_matrix.iloc[train_size:]

print(f"Training set: {len(train_returns)} days ({train_returns.index.min()} to {train_returns.index.max()})")
print(f"Validation set: {len(val_returns)} days ({val_returns.index.min()} to {val_returns.index.max()})")

# Create training environment
train_env = ESGPortfolioEnv(
    returns_data=train_returns,
    esg_scores=esg_scores,
    initial_balance=1000000,
    esg_weight=0.3,
    risk_penalty=0.5,
    transaction_cost=0.001,
    min_esg_threshold=55,
    lookback_window=20
)

# Wrap in Monitor for logging
train_env = Monitor(train_env)

# Create validation environment
val_env = ESGPortfolioEnv(
    returns_data=val_returns,
    esg_scores=esg_scores,
    initial_balance=1000000,
    esg_weight=0.3,
    risk_penalty=0.5,
    transaction_cost=0.001,
    min_esg_threshold=55,
    lookback_window=20
)

val_env = Monitor(val_env)

print("\nTraining and validation environments created")
```

---
## 6. Define Custom Callback for Training Monitoring

```python
class TrainingCallback(BaseCallback):
    """
    Custom callback for logging training progress.
    """
    def __init__(self, log_freq=1000, verbose=0):
        super(TrainingCallback, self).__init__(verbose)
        self.log_freq = log_freq
        self.episode_rewards = []
        self.episode_esg_scores = []
        self.episode_lengths = []

    def _on_step(self) -> bool:
        # Log every log_freq steps
        if self.n_calls % self.log_freq == 0:
            # Get info from environment
            if len(self.model.ep_info_buffer) > 0:
                mean_reward = np.mean([ep_info['r'] for ep_info in self.model.ep_info_buffer])
                mean_length = np.mean([ep_info['l'] for ep_info in self.model.ep_info_buffer])

                print(f"\n{'='*60}")
                print(f"Timesteps: {self.n_calls:,}")
                print(f"Mean Episode Reward: {mean_reward:.2f}")
                print(f"Mean Episode Length: {mean_length:.0f}")
                print(f"{'='*60}")

        return True

print("\nCustom callback defined")
```

---
## 7. Initialize PPO Agent

```python
print("Initializing PPO Agent...\n")

# PPO Hyperparameters
model = PPO(
    policy="MlpPolicy",
    env=train_env,
    learning_rate=3e-4,
    n_steps=1024,
    batch_size=64,
    n_epochs=5,
    gamma=0.99,
    gae_lambda=0.95,
    clip_range=0.2,
    ent_coef=0.01,
    vf_coef=0.5,
    max_grad_norm=0.5,
    verbose=1,
    tensorboard_log="./esg_portfolio_tensorboard/"
)

print("\nPPO Agent initialized")
print(f"   Policy: MLP (Multi-Layer Perceptron)")
print(f"   Learning Rate: 3e-4")
print(f"   Batch Size: 64")
print(f"   Device: {model.device}")
```

---
## 8. Train the PPO Agent

```python
print("Starting PPO Training...\n")
print("This will take approximately 30-60 minutes depending on your hardware.")
print("GPU acceleration is highly recommended!\n")

# Setup callbacks
training_callback = TrainingCallback(log_freq=5000)

# Train the agent
TOTAL_TIMESTEPS = 30000  # Adjust based on time/resources

print(f"Training for {TOTAL_TIMESTEPS:,} timesteps...\n")

model.learn(
    total_timesteps=TOTAL_TIMESTEPS,
    callback=training_callback,
    progress_bar=True
)

print("\n" + "="*60)
print("TRAINING COMPLETE!")
print("="*60)
```

## 10. Evaluate Trained Agent on Validation Set

```python
print("Evaluating trained agent on validation set...\n")

# Reset validation environment
obs, info = val_env.reset()
done = False

val_rewards = []
val_actions = []

while not done:
    # Get action from trained agent
    action, _states = model.predict(obs, deterministic=True)

    # Take step
    obs, reward, terminated, truncated, info = val_env.step(action)
    done = terminated or truncated

    val_rewards.append(reward)
    val_actions.append(action)

# Get environment history - access the unwrapped environment
portfolio_values = val_env.unwrapped.portfolio_history  # <-- Added .unwrapped
esg_scores_history = val_env.unwrapped.esg_history      # <-- Added .unwrapped
final_weights = val_env.unwrapped.portfolio_weights     # <-- Added .unwrapped

# Calculate metrics
total_return = (portfolio_values[-1] - portfolio_values[0]) / portfolio_values[0]
mean_esg = np.mean(esg_scores_history)
min_esg = np.min(esg_scores_history)

print("VALIDATION RESULTS:")
print("="*60)
print(f"Initial Portfolio Value: ${portfolio_values[0]:,.2f}")
print(f"Final Portfolio Value: ${portfolio_values[-1]:,.2f}")
print(f"Total Return: {total_return*100:.2f}%")
print(f"Mean ESG Score: {mean_esg:.2f}")
print(f"Min ESG Score: {min_esg:.2f}")
print(f"Number of Trading Days: {len(portfolio_values)-1}")
print("="*60)
```
